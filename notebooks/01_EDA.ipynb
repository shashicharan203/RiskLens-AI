# LedgerMind-AI: Notebook 01 - Exploratory Data Analysis (EDA)

This notebook analyzes the financial transaction patterns, feature distributions, credit stress metrics, and transaction risk labels.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('../data/transactions.csv')
print(f"Dataset Shape: {df.shape}")
df.head()
```

```python
df.info()
df.describe()
```

```python
# Risk Label Class Distribution
risk_counts = df['is_risk'].value_counts(normalize=True)
print("Class Proportions:", risk_counts.to_dict())

plt.figure(figsize=(6, 4))
sns.countplot(data=df, x='is_risk', palette='viridis')
plt.title('Financial Risk Label Imbalance')
plt.show()
```

```python
# Merchant Category Risk Breakdown
plt.figure(figsize=(10, 5))
sns.barplot(data=df, x='merchant_category', y='is_risk', estimator=np.mean, ci=None)
plt.xticks(rotation=45)
plt.title('Risk Probability by Merchant Category')
plt.show()
```
